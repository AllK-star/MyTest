**1. Нормализовать данные**

```python
data_norm = df#[['age', 'weight', 'height', 'ap_lo', 'ap_hi']]
data_norm = (data_norm-data_norm.mean())/(data_norm.std())
data_norm[['Family', 'Genus', 'Species', 'RecordID']] = df[['Family', 'Genus', 'Species', 'RecordID']]
data_norm.head()
```

```python
family = pd.get_dummies(data_norm["Family"], prefix="fam")
data_norm = pd.concat([data_norm, family], axis=1)
data_norm.head()
```

**2. Построить бинарный классификатор для каждого семейства(Family) на основе LogisticRegression c $l_1$ и $l_2$ регуляризацими. Это соответствует целевым исходам (Конкретное семейство, Всё остальное). Подобрать гиперпараметр $C$ с помощью кросс-валидации, используя метрику ROC AUC.**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
```

```python
features = data_norm.iloc[:, 2:24]
features.head()
```

```python
families = ['fam_Bufonidae', 'fam_Dendrobatidae', 'fam_Hylidae', 'fam_Leptodactylidae']
results = {}
for family in families:
    y_binary = data_norm[family]
    
    # Разделение на обучающую и тестовую выборки
    features_train, features_test, y_train, y_test = train_test_split(features, y_binary, test_size=0.3, random_state=0)
    
    # Подбор гиперпараметра C
    param_grid = {'C': [0.01, 0.1, 1, 10, 100]}
    
    # Модель с L1 регуляризацией
    model_l1 = GridSearchCV(LogisticRegression(penalty='l1', solver='liblinear'), param_grid, scoring='roc_auc', cv=5)
    model_l1.fit(features_train, y_train)
    y_pred_l1 = model_l1.predict_proba(features_test)[:, 1]
    auc_l1 = roc_auc_score(y_test, y_pred_l1)
    
    
    # Модель с L2 регуляризацией
    model_l2 = GridSearchCV(LogisticRegression(penalty='l2'), param_grid, scoring='roc_auc', cv=5)
    model_l2.fit(features_train, y_train)
    y_pred_l2 = model_l2.predict_proba(features_test)[:, 1]
    auc_l2 = roc_auc_score(y_test, y_pred_l2)
    
    # Сохранение результатов
    results[family] = {'AUC_L1': auc_l1, 'AUC_L2': auc_l2}
# Вывод результатов
for family, aucs in results.items():
    print(f"Family: {family}, AUC L1: {aucs['AUC_L1']}, AUC L2: {aucs['AUC_L2']}")
```

**3. Добавить полиномиальные признаки до второго порядка. Построить бинарный классификатор для каждого семейства на основе LogisticRegression c $l_1$ и $l_2$ регуляризацими. Подобрать гиперпараметр $C$ с помощью кросс-валидации, используя метрику ROC AUC.**

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2, include_bias=False)
features = poly.fit_transform(features)
```

```python
families = ['fam_Bufonidae', 'fam_Dendrobatidae', 'fam_Hylidae', 'fam_Leptodactylidae']
results = {}
for family in families:
    y_binary = data_norm[family]
    
    # Разделение на обучающую и тестовую выборки
    features_train, features_test, y_train, y_test = train_test_split(features, y_binary, test_size=0.3, random_state=0)
    
    # Подбор гиперпараметра C
    param_grid = {'C': [0.01, 0.1, 1, 10, 100]}
    
    # Модель с L1 регуляризацией
    model_l1 = GridSearchCV(LogisticRegression(penalty='l1', solver='liblinear'), param_grid, scoring='roc_auc', cv=5)
    model_l1.fit(features_train, y_train)
    y_pred_l1 = model_l1.predict_proba(features_test)[:, 1]
    auc_l1 = roc_auc_score(y_test, y_pred_l1)
    best_C_l1 = model_l1.best_params_['C']
    
    # Модель с L2 регуляризацией
    model_l2 = GridSearchCV(LogisticRegression(penalty='l2'), param_grid, scoring='roc_auc', cv=5)
    model_l2.fit(features_train, y_train)
    y_pred_l2 = model_l2.predict_proba(features_test)[:, 1]
    auc_l2 = roc_auc_score(y_test, y_pred_l2)
    best_C_l2 = model_l2.best_params_['C']
    
    # Сохранение результатов
    results[family] = {'AUC_L1': auc_l1, 'AUC_L2': auc_l2, 'Best_C_L1': best_C_l1, 'Best_C_L2': best_C_l2}
    # Вывод результатов с выбранными значениями C
for family, metrics in results.items():
    print(f"Family: {family}, AUC L1: {metrics['AUC_L1']:.4f}, Best C L1: {metrics['Best_C_L1']}, "
          f"AUC L2: {metrics['AUC_L2']:.4f}, Best C L2: {metrics['Best_C_L2']}")
```
